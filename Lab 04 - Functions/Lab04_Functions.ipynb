## 1. Why Functions Exist

The same averaging logic, written three times and then written once.

```python
# without a function - the logic is copied three times
ali, sara, bilal = [78, 45, 90], [88, 92, 95], [65, 70, 58]

print("Ali  :", round(sum(ali) / len(ali), 2))
print("Sara :", round(sum(sara) / len(sara), 2))
print("Bilal:", round(sum(bilal) / len(bilal), 2))
```

## 3. Parameters and Arguments

The names in the definition are **parameters**; the values at the call are **arguments**.

### 3.1 Positional — order carries the meaning

```python
def power(base, exponent):
    return base ** exponent

print(power(2, 10))
print(power(10, 2))      # same arguments, different meaning
```

### 3.2 Keyword — the name carries the meaning

```python
print(power(exponent=10, base=2))    # order no longer matters
print(power(2, exponent=10))         # positional first, then keyword

# power(base=2, 10)   -> SyntaxError: positional argument follows keyword argument
```

### 3.3 Default values

```python
def average(marks, ndigits=2):
    return round(sum(marks) / len(marks), ndigits)

print(average([78, 45, 91]))             # default ndigits = 2
print(average([78, 45, 91], 0))          # positionally
print(average([78, 45, 91], ndigits=4))  # by keyword
```

```python
def greet(name, greeting="Assalam-o-Alaikum", punctuation="!"):
    return greeting + ", " + name + punctuation

print(greet("Ali"))
print(greet("Sara", "Good morning"))
print(greet("Bilal", punctuation="."))
```

### 3.4 The mutable default trap

A default is evaluated **once**, when the function is defined. A list default is therefore
shared by every call that relies on it.

```python
def add_mark(mark, marks=[]):      # dangerous
    marks.append(mark)
    return marks

print(add_mark(50))
print(add_mark(60))                # the same list is still there
print(add_mark(70))
```

```python
def add_mark(mark, marks=None):    # correct
    if marks is None:
        marks = []
    marks.append(mark)
    return marks

print(add_mark(50))
print(add_mark(60))
```

****kwargs:** Could take infinite amount of keyword arguments and create as tuple. **Example**: kwargs={'name':'Ali', 'age':21, ...} -> dictionary

```python
def total(*marks):
    print(type(marks), marks)
    return sum(marks)

print(total(78, 45))
print(total(78, 45, 90, 62))
print(total())
```

```python
# the * at a CALL does the opposite job: it spreads a sequence into separate arguments
values = [2, 10]
print(power(*values))            # same as power(2, 10)

settings = {'base': 3, 'exponent': 4}
print(power(**settings))         # ** spreads a dictionary into keyword arguments
```

## 4. Return Values

### 4.1 `return` is not `print`

`print()` shows characters to a human. `return` hands a **value** back to the code that made
the call. A function that prints instead of returning cannot be reused in a calculation.

```python
def average_printed(marks):
    print(sum(marks) / len(marks))     # shows the answer, returns nothing

def average_returned(marks):
    return sum(marks) / len(marks)     # hands the answer back

a = average_printed([78, 45, 90])
b = average_returned([78, 45, 90])

print("a =", a)                        # None - there was no return
print("b =", b)
print("b doubled:", b * 2)
```

```python
# what happens if you try to use the value that was only printed
try:
    print(a * 2)
except TypeError as err:
    print("TypeError:", err)
```

### 4.2 Returning several values (they arrive as a tuple)

```python
def statistics(marks):
    return len(marks), sum(marks), sum(marks) / len(marks)

count, total_marks, mean = statistics([78, 45, 90, 62])
print(count, total_marks, round(mean, 2))

result = statistics([78, 45])          # not unpacked
print(result, type(result))
```

### 4.3 `return` ends the function immediately

```python
def safe_average(marks):
    if not marks:            # an empty list is false
        return 0             # leave at once - nothing below runs
    return sum(marks) / len(marks)

print(safe_average([78, 45, 90]))
print(safe_average([]))      # no ZeroDivisionError
```

```python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False     # one divisor is enough to decide
    return True              # only reached if the loop finished

print([n for n in range(1, 20) if is_prime(n)])
```

## 5. Scope

Every call creates a fresh local workspace holding the parameters and any variable created in
the body. When the call ends, that workspace is thrown away. A name is looked up
**Local → Enclosing → Global → Built-in**.

```python
def average(marks):
    total = sum(marks)       # local to this call
    return total / len(marks)

print(average([78, 45, 90]))

try:
    print(total)             # the local name does not exist out here
except NameError as err:
    print("NameError:", err)
```

### 5.1 Reading a global is fine

```python
PASSING_MARK = 50            # a global constant

def has_passed(mark):
    return mark >= PASSING_MARK

print(has_passed(45), has_passed(78))
```

### 5.2 Assigning makes a name local — for the whole body

```python
count = 0

def register():
    count = count + 1        # Python treats count as local, so reading it fails
    return count

try:
    register()
except UnboundLocalError as err:
    print("UnboundLocalError:", err)
```

```python
# way out 1: the global keyword - works, but avoid it
count = 0

def register_global():
    global count
    count = count + 1
    return count

register_global(); register_global()
print("count is now", count)
```

```python
# way out 2 (preferred): take the value in, hand the new value back
def register(value):
    return value + 1

count = 0
count = register(count)
count = register(count)
print("count is now", count)
```

### 5.3 Shadowing — a parameter can hide a global of the same name

```python
marks = [78, 45, 90]         # global

def rescale(marks):          # this parameter shadows the global
    marks = [m + 5 for m in marks]
    return marks

print(rescale(marks))
print(marks)                 # the global list is untouched
```

### 5.4 Can a function change its argument?

Arguments are passed by object reference. Numbers, strings and tuples are immutable, so the
caller can never see a change. Lists, dictionaries and sets are mutable, so a method call
inside the function **is** visible outside.

```python
def try_to_change(number, items):
    number = number + 100    # rebinds the local name only
    items.append(100)        # modifies the caller's list

n = 1
L = [1]
try_to_change(n, L)
print("n =", n)              # unchanged
print("L =", L)              # changed
```

```python
# decide deliberately which kind of function you are writing
def add_bonus_copy(marks, bonus):
    """Return a NEW list; the caller's list is untouched."""
    return [m + bonus for m in marks]

def add_bonus_inplace(marks, bonus):
    """Modify the caller's list; return None (like list.sort does)."""
    for i in range(len(marks)):
        marks[i] += bonus

original = [78, 45, 90]
print("copy    :", add_bonus_copy(original, 5), "| original:", original)

add_bonus_inplace(original, 5)
print("in place:", original)
```

## 7. `zip`, `enumerate`, and Passing a Function to a Function

You used `zip()` and `enumerate()` as loop helpers in Lab 02. They are ordinary functions that
return a **lazy** object — it produces items as you loop, and it is exhausted after one pass.

```python
subjects = ['Physics', 'Chemistry', 'Maths']
marks    = [78, 45, 90]

print(zip(subjects, marks))                 # a lazy object, not a list
print(list(zip(subjects, marks)))
print(dict(zip(subjects, marks)))
print(list(enumerate(subjects, start=1)))
```

## 8. `lambda`

A function written as a single expression, with no name and no `return` — the expression *is*
the result. Use it where a function is needed once, as an argument to something else.

```python
def double(x):
    return x * 2

double_lambda = lambda x: x * 2      # equivalent
area = lambda length, width: length * width

print(double(21), double_lambda(21), area(3, 4))
```

```python
students = [('Ali', 78), ('Sara', 92), ('Bilal', 65)]

print(sorted(students, key=lambda pair: pair[1]))                # by marks
print(sorted(students, key=lambda pair: pair[1], reverse=True))  # best first
print(max(students, key=lambda pair: pair[1]))

records = [{'name': 'Ali', 'cgpa': 3.4}, {'name': 'Sara', 'cgpa': 3.9}]
print(sorted(records, key=lambda r: r['cgpa'], reverse=True))
```

**Style note.** `double = lambda x: x * 2` gives you every limitation of a lambda and none of
the benefits of a `def` — no docstring, no statements, a worse traceback. If it deserves a
name, write `def`.

## 9. `map()` and `filter()`

`map(f, L)` applies `f` to every item. `filter(p, L)` keeps the items for which `p` is true.
Both are lazy — wrap them in `list()` to see the result.

```python
marks = [45, 78, 32, 90, 61]

passed  = list(filter(lambda m: m >= 50, marks))
bumped  = list(map(lambda m: m + 5, passed))
squares = list(map(lambda n: n ** 2, [1, 2, 3, 4]))

print(passed)
print(bumped)
print(squares)
```

```python
# map over two sequences at once - one argument taken from each
obtained = [78, 45, 90]
totals   = [100, 50, 100]

percentages = list(map(lambda got, out_of: round(got / out_of * 100, 1),
                       obtained, totals))
print(percentages)
```

### 9.1 `map` / `filter` or a comprehension?

| Task | With map / filter | As a comprehension |
|---|---|---|
| Transform every item | `list(map(f, L))` | `[f(x) for x in L]` |
| Keep some items | `list(filter(p, L))` | `[x for x in L if p(x)]` |
| Both at once | `list(map(f, filter(p, L)))` | `[f(x) for x in L if p(x)]` |

The comprehension usually reads better; learn both, because `map` and `filter` are everywhere
in existing code and in the data libraries later in this course.

```python
marks = [45, 78, 32, 90, 61]

print(list(map(lambda m: m + 5, filter(lambda m: m >= 50, marks))))
print([m + 5 for m in marks if m >= 50])       # identical, easier to read
```

## 10. Putting It Together — A Small Marks Module

Several short functions, each with one job and a docstring, and a `main()` that only calls
them in order. Only `report()` prints — that is its job. This is the shape your Class Task
solutions should take.

```python
def average(marks, ndigits=2):
    """Return the mean of a list of marks."""
    if not marks:
        return 0
    return round(sum(marks) / len(marks), ndigits)


def grade(mark):
    """Return the letter grade for one mark."""
    if mark >= 80:
        return 'A'
    elif mark >= 70:
        return 'B'
    elif mark >= 60:
        return 'C'
    elif mark >= 50:
        return 'D'
    return 'F'


def passed(records):
    """Return only the records whose average is at least 50."""
    return list(filter(lambda r: average(r['marks']) >= 50, records))


def ranked(records):
    """Return the records sorted from the highest average to the lowest."""
    return sorted(records, key=lambda r: average(r['marks']), reverse=True)


def report(records):
    """Print one line per student, best first."""
    for position, record in enumerate(ranked(records), start=1):
        avg = average(record['marks'])
        print(position, record['name'], avg, grade(avg))


def main():
    students = [
        {'name': 'Ali',   'marks': [78, 45, 90]},
        {'name': 'Sara',  'marks': [88, 92, 95]},
        {'name': 'Bilal', 'marks': [40, 38, 52]},
    ]
    report(students)
    print("passed:", [r['name'] for r in passed(students)])
    print("class average:", average([average(r['marks']) for r in students]))


main()
```

### C2 — Flexible Arguments
*defaults · `*args` · `**kwargs`*

Write `make_result_card(name, roll, *marks, passing=50, **details)` which **returns** (does not
print) a multi-line string containing:

* the student's name and roll number
* every mark passed in, numbered with `enumerate()`
* the total, the average, and how many marks were below `passing`
* the verdict `"PASS"` or `"FAIL"`
* any extra fields supplied as keyword arguments, one per line

Call it at least four times to show every feature — with two marks, with five marks, with
`passing` overridden, and with two extra details such as `section` and `semester` — printing
each returned string.

```python
# C2 - Flexible Arguments
# Name: Sarim Raza Ansari
# Roll number: 25k0064

# Your code here
def make_result_card(name, roll, *marks, passing=50, **details):
    """Result card for a student."""
    total = sum(marks)
    average = total / len(marks) if marks else 0
    below = sum(1 for mark in marks if mark < passing)
    verdict = "PASS" if marks and below == 0 else "FAIL"

    lines = [
        f"Name: {name}",
        f"Roll: {roll}"
    ]

    for number, mark in enumerate(marks, start=1):
        lines.append(f"Mark {number}: {mark}")

    lines.append(f"Total: {total}")
    lines.append(f"Average: {average:.2f}")
    lines.append(f"Below {passing}: {below}")
    lines.append(f"Verdict: {verdict}")

    for field, value in details.items():
        lines.append(f"{field}: {value}")

    return "\n".join(lines)

print(make_result_card("Ali", "25K0001", 78, 85))
print(make_result_card("Sarim", "25K0002", 88, 92, 95, 76, 81))
print(make_result_card("Kurosaki", "25K0003", 45, 55, 60, passing=60))
print(make_result_card("Unahana", "25K0004", 72, 81, 90, section="A", semester=2))

```

```python
# C5 - A Text Toolkit
# Name: Sarim Raza Ansari
# Roll number: 25k0064

# Your code here
# assert is_palindrome('level')
# assert not is_palindrome('python')
# assert word_count('a a b')['a'] == 2

def clean(text):
    punctuation = ".,!?;:'\"-()[]{}"
    text = text.lower()

    for char in punctuation:
        text = text.replace(char, "")

    return text


def word_count(text):
    words = clean(text).split()
    counts = {}

    for word in words:
        counts[word] = counts.get(word, 0) + 1

    return counts


def top_n(counts, n=3):
    return sorted(counts.items(), key=lambda pair: pair[1], reverse=True)[:n]


def is_palindrome(word):
    word = clean(word)
    return word == word[::-1]


def report(text):
    """Print a summary of the text."""
    cleaned = clean(text)
    counts = word_count(text)
    top = top_n(counts)

    print("Cleaned text:", cleaned)
    print("Word count:", counts)
    print("Top words:", top)


assert clean("Hello, WORLD!") == "hello world"
assert word_count("a a b") == {"a": 2, "b": 1}
assert top_n({"a": 3, "b": 1, "c": 2}) == [("a", 3), ("c", 2), ("b", 1)]
assert is_palindrome("level") == True
assert is_palindrome("python") == False

text = input("Enter a sentence: ")
report(text)
```
